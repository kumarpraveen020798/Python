# 22. String.istitle()

Returns True if each word in the string starts with an uppercase letter followed by lowercase letters. Otherwise, it returns False
Every word must follow the Title Case format ,
Digits and special characters do not affect the result ,
The string must contain at least one alphabetic character.

```python
text = "Python Programming"
# Both words start with an uppercase letter.
# The remaining letters are lowercase.
# Therefore, istitle() returns True.
print(text.istitle())

text = "PYTHON PROGRAMMING"
# Every letter is uppercase.
# The remaining letters after the first
# should be lowercase.
# Therefore, istitle() returns False.
print(text.istitle())

text = "Python 3"
# "Python" is in Title Case.
# "3" is a digit and does not affect the result.
# Therefore, istitle() returns True.
print(text.istitle())
```

# 24. separator.join(iterable)

Joins all the elements of an iterable (such as a list, tuple, or set) into a single string, using the specified separator.

```python
fruits = ["Apple", "Banana", "Mango"]
# "," is used as the separator.
# It is inserted between every element.
result = ",".join(fruits)
print(result)

words = ["Python", "is", "awesome"]
# A single space is used as the separator.
# All words become one sentence.
result = " ".join(words)
print(result)
```

# 25. String.ljust(width[, fillchar])

Returns a left-justified string of the specified width. If the original string is shorter than the given width, it is padded on the right side with spaces (or the specified fill character).

```python
text = "Python"
# The length of "Python" is 6.
# The required width is 10.
# Therefore, 4 spaces are added to the right.
result = text.ljust(10)
print(result)

text = "Python"
# The total width is 10.
# '*' is used as the fill character.
# Four '*' characters are added to the right.
result = text.ljust(10, "*")
print(result)
```

# 27. String.lstrip([chars])

Removes leading characters (characters from the left side) of a string. By default, it removes whitespace characters such as spaces, tabs, and newlines.

```python
text = "    Python"
# Leading spaces are removed.
# The word "Python" remains unchanged.
result = text.lstrip()
print(result)

text = "###Python"
# '#' is specified in chars.
# All leading '#' characters are removed.
result = text.lstrip("#")
print(result)

text = "abcabcPython"
# The characters 'a', 'b', and 'c'
# are removed from the left side
# until a different character is found.
result = text.lstrip("abc")
print(result)
```

# 28. String.maketrans(x[, y[, z]])

Creates a translation table that is used with the translate() method to replace or remove characters from a string.

```python
text = "apple"
# Replace:
# a -> x
# p -> y
# l -> z
table = str.maketrans("apl", "xyz")
result = text.translate(table)
print(result)

text = "python"
# Replace:
# p -> P
# y -> Y
# t -> T
# h -> H
# o -> O
# n -> N
table = str.maketrans("python", "PYTHON")
print(text.translate(table))

text = "cat dog"
# Dictionary keys are characters.
# c becomes C
# d becomes D
table = str.maketrans({
    "c": "C",
    "d": "D"
})
print(text.translate(table))
```

# 29. String.partition(separator)

Splits a string into three parts using the first occurrence of the specified separator.

```python
text = "Python Programming"
# The first space is used as the separator.
# The string is divided into three parts.
result = text.partition(" ")
print(result)

text = "2026-07-31"
# The first '-' is found.
# The string is split into three parts.
result = text.partition("-")
print(result)

text = "Python"
# ',' is not present in the string.
# Therefore, the original string is returned
# along with two empty strings.
result = text.partition(",")
print(result)
```

# 33. String.rfind(substring[, start[, end]])

Returns the highest (rightmost) index of the specified substring in a string.
It searches from right to left, but the returned index is still counted from the beginning (left side) of the string.


```python
text = "Python is easy. Python is powerful."
# Find the last occurrence of "Python".
# The search starts from the right.
result = text.rfind("Python")
print(result)

text = "Python Programming"
# "Java" is not present.
# Therefore, -1 is returned.
result = text.rfind("Java")
print(result)

text = "banana"
# Search between index 0 and 5.
# The character at index 5
# is NOT included because
# end is exclusive.
result = text.rfind("a", 0, 5)
print(result)
```

# 37. string.rsplit(separator=None, maxsplit=-1)

Splits a string into a list of substrings, starting the split from the right (end) of the string.

```python
text = "Python Programming Language"
# No separator is provided.
# The string is split on spaces.
result = text.rsplit()
print(result)

text = "A-B-C-D"
# Split using '-'.
# Since maxsplit is not specified,
# all occurrences are split.
result = text.rsplit("-")
print(result)

text = "one two three four"
# Perform only one split
# starting from the right.
result = text.rsplit(" ", 1)
print(result)
```

# 43. String.swapcase()

Returns a new string by converting:

Uppercase letters to lowercase
Lowercase letters to uppercase

All non-alphabetic characters (digits, spaces, punctuation, symbols, etc.) remain unchanged.

```python
text = "Python"
# 'P' becomes 'p'.
# 'ython' becomes 'YTHON'.
result = text.swapcase()
print(result)

text = "PYTHON"
# Every uppercase letter
# becomes lowercase.
result = text.swapcase()
print(result)

text = "PyThOn PrOgRaMmInG"
# Every uppercase letter becomes lowercase.
# Every lowercase letter becomes uppercase.
result = text.swapcase()
print(result)
```

# 45. String.translate(table)

Returns a new string by replacing or removing characters according to a translation table.

```python
text = "apple"
# Create a translation table.
# Replace:
# a -> A
# p -> P
table = str.maketrans({
    "a": "A",
    "p": "P"
})
result = text.translate(table)
print(result)

text = "Programming"
# Remove all vowels.
table = str.maketrans("", "", "aeiou")
result = text.translate(table)
print(result)
```

# 47. String.zfill(width)

Returns a new string by padding zeros (0) to the left of the string until it reaches the specified width.

```python
text = "25"
# The current length is 2.
# The required width is 5.
# Therefore, three zeros are added
# to the left.
result = text.zfill(5)
print(result)

text = "Python"
# The string length is already 6.
# Since the width is also 6,
# no zeros are added.
result = text.zfill(6)
print(result)
```
